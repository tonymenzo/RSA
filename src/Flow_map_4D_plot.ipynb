```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# -- Load data --
plot_nm = '2'
gradients     = np.load(f'temp_results/Flow_map/4D/gradients_{plot_nm}.npy')
loss_grid     = np.load(f'temp_results/Flow_map/4D/loss_grid_{plot_nm}.npy')
mu_sigmamu    = np.load(f'temp_results/Flow_map/4D/mu_{plot_nm}.npy')
Neff          = np.load(f'temp_results/Flow_map/4D/Neff_{plot_nm}.npy')
params_grid   = np.load(f'temp_results/Flow_map/4D/params_grid_{plot_nm}.npy')  # shape: (N, 4)
ordered_params = np.load(f'temp_results/Flow_map/4D/ordered_params_{plot_nm}.npy')

# -- Plot config --
fixed_param_index = 3  # fix the 4th param (index in ordered_params)
fixed_param_name  = ordered_params[fixed_param_index]
unique_fixed_vals = np.unique(params_grid[:, fixed_param_index])

# -- Plot settings --
scale = 1e3
scatter_metric = np.log10(loss_grid)  # or Neff
num_slices = min(len(unique_fixed_vals), 4)

# Loop over a few slices
fig = plt.figure(figsize=(6 * num_slices, 6))
for i, val in enumerate(unique_fixed_vals[:num_slices]):
    ax = fig.add_subplot(1, num_slices, i + 1, projection='3d')

    # Get 3D slice where the fixed param = val
    mask = params_grid[:, fixed_param_index] == val
    sliced_params = params_grid[mask]
    sliced_grads = -gradients[mask]  # minus gradient = descent direction
    sliced_loss = scatter_metric[mask]

    # Determine indices for the 3 varying params
    varying_indices = [j for j in range(4) if j != fixed_param_index]
    X, Y, Z = sliced_params[:, varying_indices[0]], sliced_params[:, varying_indices[1]], sliced_params[:, varying_indices[2]]
    U, V, W = sliced_grads[:, varying_indices[0]] / scale, sliced_grads[:, varying_indices[1]] / scale, sliced_grads[:, varying_indices[2]] / scale

    # Plot quiver
    ax.quiver(X, Y, Z, U, V, W, length=1.0, normalize=False)

    # Scatter colored by loss or Neff
    sc = ax.scatter(X, Y, Z, c=sliced_loss, cmap='plasma', s=20, alpha=0.8)
    
    ax.set_xlabel(ordered_params[varying_indices[0]])
    ax.set_ylabel(ordered_params[varying_indices[1]])
    ax.set_zlabel(ordered_params[varying_indices[2]])
    ax.set_title(f'{fixed_param_name} = {val:.3f}')

cbar = fig.colorbar(sc, ax=fig.get_axes(), shrink=0.6, location='right')
cbar.set_label('log10(loss)')
fig.suptitle(f'3D Slices of 4D Gradient Field (plot {plot_nm})', fontsize=14)
plt.tight_layout()
plt.savefig(f'/pscratch/sd/l/ljpuslar/RSA/RSA/src/temp_results/Flow_map/4D/images/1_{plot_nm}.png')
plt.show()
```

```python
import numpy as np
import os
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# -- Load data --
plot_nm = '2'
base_dir = '/pscratch/sd/l/ljpuslar/RSA/RSA/src/temp_results/Flow_map/4D'
img_dir  = os.path.join(base_dir, 'images')
os.makedirs(img_dir, exist_ok=True)

gradients      = np.load(f'{base_dir}/gradients_{plot_nm}.npy')      # shape: (N, 4)
loss_grid      = np.load(f'{base_dir}/loss_grid_{plot_nm}.npy')      # shape: (N,)
mu_sigmamu     = np.load(f'{base_dir}/mu_{plot_nm}.npy')             # shape: (N, 2)
Neff           = np.load(f'{base_dir}/Neff_{plot_nm}.npy')           # shape: (N,)
params_grid    = np.load(f'{base_dir}/params_grid_{plot_nm}.npy')    # shape: (N, 4)
ordered_params = np.load(f'{base_dir}/ordered_params_{plot_nm}.npy') # e.g. ['a', 'b', 'sigma', 'c']

# Prepare the scalar metric to color by (log10 loss here)
scatter_metric = np.log10(loss_grid)

# Scaling factor for arrows
scale = 1e3

# For each of the 4 parameters, fix it and plot the remaining 3D slice
num_dims = params_grid.shape[1]
for fixed_dim in range(num_dims):
    fixed_name = ordered_params[fixed_dim]
    unique_vals = np.unique(params_grid[:, fixed_dim])
    n_slices = min(len(unique_vals), 4)

    # Create a wide figure to host multiple side-by-side 3D plots
    fig = plt.figure(figsize=(6 * n_slices, 6))
    for i, val in enumerate(unique_vals[:n_slices]):
        ax = fig.add_subplot(1, n_slices, i + 1, projection='3d')
        
        # Mask for this slice
        mask = params_grid[:, fixed_dim] == val
        pts = params_grid[mask]
        grads = -gradients[mask]  # descent direction
        col  = scatter_metric[mask]

        # Identify the three varying dimensions
        vary_dims = [d for d in range(num_dims) if d != fixed_dim]
        X = pts[:, vary_dims[0]]
        Y = pts[:, vary_dims[1]]
        Z = pts[:, vary_dims[2]]
        U = grads[:, vary_dims[0]] / scale
        V = grads[:, vary_dims[1]] / scale
        W = grads[:, vary_dims[2]] / scale

        # Quiver and scatter
        ax.quiver(X, Y, Z, U, V, W, length=1.0, normalize=False)
        sc = ax.scatter(X, Y, Z, c=col, cmap='plasma', s=20, alpha=0.8)

        # Labels and title
        ax.set_xlabel(ordered_params[vary_dims[0]])
        ax.set_ylabel(ordered_params[vary_dims[1]])
        ax.set_zlabel(ordered_params[vary_dims[2]])
        ax.set_title(f"{fixed_name} = {val:.3f}")

    # Shared colorbar on the right
    cbar = fig.colorbar(sc, ax=fig.get_axes(), shrink=0.6, location='right')
    cbar.set_label('log10(loss)')

    fig.suptitle(f'3D Slices of 4D Flow (fixing {fixed_name}) – plot {plot_nm}', fontsize=14)
    plt.tight_layout(rect=[0, 0, 0.95, 0.95])

    # Save each figure named by the fixed parameter
    output_path = os.path.join(img_dir, f'{fixed_name}_slices_{plot_nm}.png')
    # plt.savefig(output_path, dpi=300)
    # plt.close()
```
