```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2
from matplotlib.patches import Ellipse

# =========================
# 1️⃣ Configuration
# =========================
observable = 'Joker'
observable = 'Joker_locked'
run_nm     = '7'
run_nm     = '3'

plot_configs = {
    'Joker': {
        '7': {
            'data_folder': '/pscratch/sd/l/ljpuslar/RSA/RSA/src/temp_results/Tuner_Confidence_ND/Joker/2D/',
            'xlabel': r'$a_d$',
            'ylabel': r'$b_d$',
            'title':  'RSA Tuning — Joker (2D)',
            'color':  'tab:red',
            'marker': 'o',
            'size':   25,
            # Base & target points
            'base_point':   (0.68 + 0.1275, 0.905),
            'target_point': (0.68, 0.98),
        },
    },
    'Joker_locked': {
        '2': {
            'data_folder': '/pscratch/sd/l/ljpuslar/RSA/RSA/src/temp_results/Tuner_Confidence_ND/Joker/locked/6D/',
            'xlabel': r'$a_d$',
            'ylabel': r'$b_d$',
            'title':  'RSA Tuning — Joker (2D)',
            'color':  'tab:red',
            'marker': 'o',
            'size':   25,
            # Base & target points
            'base_point':   (0.68 + 0.04, 0.88),
            'target_point': (0.68, 0.98),
        },
    },
    'Joker_locked': {
        '3': {
            'data_folder': '/pscratch/sd/l/ljpuslar/RSA/RSA/src/temp_results/Tuner_Confidence_ND/Joker/locked/6D/',
            'xlabel': r'$a_d$',
            'ylabel': r'$b_d$',
            'title':  'RSA Tuning — Joker (2D)',
            'color':  'tab:red',
            'marker': 'o',
            'size':   25,
            # Base & target points
            'base_point':   (0.68 + 0.04, 0.88),
            'target_point': (0.68, 0.98),
        },
    },
}

def get_config(obs, run):
    try:
        return plot_configs[obs][run]
    except KeyError:
        raise ValueError(f"No config found for observable='{obs}', run='{run}'")

config = get_config(observable, run_nm)
data_folder = config['data_folder']

# =========================
# 2️⃣ Load data
# =========================
points = np.load(os.path.join(data_folder, f'all_params_{run_nm}.npy'))
loss_values = np.load(os.path.join(data_folder, f'loss_values_{run_nm}.npy'))

if points.ndim == 2:
    points = points[None, :, :]
n_traj, n_steps, _ = points.shape

print(f"Loaded points: {points.shape}")
print(f"Loaded losses: {loss_values.shape}")

# =========================
# 3️⃣ Plot setup
# =========================
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(6, 6))

# Color gradient across all points
colors = plt.cm.plasma(np.linspace(0.1, 0.9, n_traj))

# ---- Scatter trajectories ----
xs, ys = [], []
for i in range(n_traj):
    x, y = points[i, -1, 0], points[i, -1, 1]
    # print(len(x))
    # Alpha fades slightly with step
    # alphas = np.linspace(0.3, 0.9, len(x))
    # ax.scatter(
    #     x, y,
    #     s=config['size'],
    #     c='blue',
    #     alpha=0.5,
    #     edgecolors='none',
    #     marker=config['marker'],
    #     zorder=1
    # )
    xs.append(x)
    ys.append(y)

# Confidence ellipse:
data = np.vstack((xs, ys))

# Covariance and mean
cov = np.cov(data)
mean = np.mean(data, axis=1)

# Eigenvalues and eigenvectors
vals, vecs = np.linalg.eigh(cov)

# Sort by largest eigenvalue
order = vals.argsort()[::-1]
vals, vecs = vals[order], vecs[:, order]

# 95% confidence level
chi2_val = chi2.ppf(0.95, df=2)
width, height = 2 * np.sqrt(vals * chi2_val)  # full lengths
angle = np.degrees(np.arctan2(*vecs[:,0][::-1]))

# Plot
# fig, ax = plt.subplots()
ax.scatter(xs, ys, s=10, alpha=0.5, label='Parameter fits')
ellipse = Ellipse(xy=mean, width=width, height=height, angle=angle,
                  edgecolor='red', facecolor='none',linestyle='--', lw=2, label='95% confidence ellipse')
ax.add_patch(ellipse)

ax.set_aspect('equal')

# ---- Key points ----
bx, by = config.get('base_point', (None, None))
tx, ty = config.get('target_point', (None, None))
if bx is not None:
    ax.scatter(bx, by, c='blue', marker='s', s=90, edgecolors='k', linewidths=0.5, label='Base', zorder=3)
if tx is not None:
    ax.scatter(tx, ty, c='magenta', marker='P', s=100, edgecolors='k', linewidths=0.5, label='Target', zorder=4)

# ---- Min-loss point ----
min_idx = np.unravel_index(np.argmin(loss_values), loss_values.shape)
min_point = points[min_idx[0], min_idx[1]]
ax.scatter(min_point[0], min_point[1], c='yellow', s=120, marker='*',
           edgecolors='k', linewidths=0.6, label='Min loss', zorder=5)

# ---- Formatting ----
ax.set_title(config['title'], fontsize=13, weight='bold')
ax.set_xlabel(config['xlabel'], fontsize=12)
ax.set_ylabel(config['ylabel'], fontsize=12)
ax.legend(frameon=True, loc='best', fontsize=9)
ax.set_aspect('equal', adjustable='box')

# Aesthetic tweaks
ax.tick_params(axis='both', which='major', labelsize=10)
ax.grid(True, linestyle='--', alpha=0.4)
for spine in ax.spines.values():
    spine.set_alpha(0.4)

plt.tight_layout()

# =========================
# 4️⃣ Save & show
# =========================
out_dir = os.path.join(data_folder, 'plots')
os.makedirs(out_dir, exist_ok=True)
out_path = os.path.join(out_dir, f'{observable}_scatter_{run_nm}.png')
plt.savefig(out_path, dpi=300, bbox_inches='tight', transparent=True)
plt.show()

print(f"✅ Scatter plot saved to: {out_path}")
```
